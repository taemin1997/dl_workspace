```python
def single_layer_perceptron(x1, x2, w1, w2, b):
    """
    단층 퍼셉트론 연산 함수
    x1, x2 : 입력값
    w1, w2 : 가중치 (Weight)
    b      : 편향 (Bias)
    """
    # 1. 입력값과 가중치의 곱에 편향을 더함 (순입력 총합)
    total_sum = (x1 * w1) + (x2 * w2) + b
    
    # 2. 활성화 함수 (임계값 0을 기준으로 한 계단 함수 적용)
    if total_sum > 0:
        return 1
    else:
        return 0

# 진리표 테스트를 위한 4가지 입력 조합
inputs = [(0, 0), (1, 0), (0, 1), (1, 1)]

# --- [1] AND 게이트 테스트 ---
print("--- [AND 게이트 진리표] ---")
w1, w2, b = 0.5, 0.5, -0.7
for x1, x2 in inputs:
    output = single_layer_perceptron(x1, x2, w1, w2, b)
    print(f"입력: ({x1}, {x2}) -> 출력: {output}")

print("\n--- [OR 게이트 진리표] ---")
# --- [2] OR 게이트 테스트 ---
w1, w2, b = 0.5, 0.5, -0.2
for x1, x2 in inputs:
    output = single_layer_perceptron(x1, x2, w1, w2, b)
    print(f"입력: ({x1}, {x2}) -> 출력: {output}")

print("\n--- [NAND 게이트 진리표] ---")
# --- [3] NAND 게이트 테스트 ---
w1, w2, b = -0.5, -0.5, 0.7
for x1, x2 in inputs:
    output = single_layer_perceptron(x1, x2, w1, w2, b)
    print(f"입력: ({x1}, {x2}) -> 출력: {output}")
```

```python
# 1. 기본 단층 퍼셉트론 엔진 구조 정의
def perceptron(x1, x2, w1, w2, b):
    total_sum = (x1 * w1) + (x2 * w2) + b
    return 1 if total_sum > 0 else 0

# 2. 앞서 검증된 매개변수(Weight, Bias)로 기본 게이트 정의
def NAND_gate(x1, x2):
    return perceptron(x1, x2, w1=-0.5, w2=-0.5, b=0.7)

def OR_gate(x1, x2):
    return perceptron(x1, x2, w1=0.5, w2=0.5, b=-0.2)

def AND_gate(x1, x2):
    return perceptron(x1, x2, w1=0.5, w2=0.5, b=-0.7)

# 3. [핵심] 게이트를 조합한 다층 퍼셉트론 XOR 구조 구현
def XOR_gate(x1, x2):
    # 은닉층 (Hidden Layer) 단계
    s1 = NAND_gate(x1, x2)  # 뉴런 1 활성화
    s2 = OR_gate(x1, x2)    # 뉴런 2 활성화
    
    # 출력층 (Output Layer) 단계
    y = AND_gate(s1, s2)    # 뉴런 3 최종 출력
    return y

# --- XOR 다층 퍼셉트론 진리표 검증 테스트 ---
print("--- [다층 퍼셉트론 XOR 게이트 검증] ---")
inputs = [(0, 0), (1, 0), (0, 1), (1, 1)]

for x1, x2 in inputs:
    output = XOR_gate(x1, x2)
    print(f"입력: ({x1}, {x2}) ──> 최종 출력: {output}")
```
